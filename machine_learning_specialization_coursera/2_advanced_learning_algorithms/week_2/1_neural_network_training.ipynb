# Neural Network Training

## Loss and functions




### Logistic Loss

- Logistic loss is also known as "Binary Crossentropy".



In Tensor Flow. Logistic loss can be utilized during compilation:

`model.compile(loss=BinaryCrossentropy())`

### Regression loss function

- Reminder: Used when prediction numbers and not categories.
- Also known as "Mean Squared Error"



`model.compile(loss=MeanSquaredError())`
