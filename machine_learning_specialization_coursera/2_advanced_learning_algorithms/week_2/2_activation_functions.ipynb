# Activation Functions

## ReLU Function

Stands for "Rectified Linear Unit".

- All values that are 0 or under, become 0.
- All values over 0 remain.
- ReLU function is the most commonly used activation function.
- Gradient descent perform much better when using ReLU compared to Sigmoid.



##

## Linear Activation Function === No Activation Function

If $g(z) = z$, then we are effectively not using an activation function. This is because the input is not modified whatsover, so the $g()$ is basically non-existent.



## Picking Activation Functions

### Output layer:
- Sigmoid: for binary classification problem.
- ReLU: when `y` can only be zero or positive values.
- Linear Regression: when `y` can be positive or negative.

### Hidden layers:

He recommends to use ReLU by default for hidden layers. RelU is considered to be faster than Sigmoid.

## Why are Activation functions needed?

If every node in a neural network used `linear regression`, then the entire network would act as a single linear regression model.
- Note: Linear regression is considered to be "none activation" because $g(z) = z$.
