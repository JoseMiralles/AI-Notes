# Activation Functions

## ReLU Function

Stands for "Rectified Linear Unit".

- All values that are 0 or under, become 0.
- All values over 0 remain.
- ReLU function is the most commonly used activation function.
- Gradient descent perform much better when using ReLU compared to Sigmoid.



## Picking Activation Functions

### Output layer:
- Sigmoid: for binary classification problem.
- Regression: when `y` can be positive or negative.
- ReLU: when `y` can only be zero or positive values.

### Hidden layers:

He recommends to use ReLU by default for hidden layers.
