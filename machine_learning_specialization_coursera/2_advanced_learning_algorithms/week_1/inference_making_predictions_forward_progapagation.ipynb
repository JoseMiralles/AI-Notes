# Digit regonition example

Goal: to recognize a number drawn in an 8x9 grid.



## Calculating probability that this is a "1"

This example neural network would be used to calculate the probability for a single number, "1".



- Layer one, $\overrightarrow a[1]$, consists of 25 units $[ \overrightarrow a ^ {[1]}_1 ... \overrightarrow a ^ {[1]}_{25} ]$.
- Layer two consists of 15 units.
- Last layer outputs the probability that the give grid makes up the drawing of the character "1".

## Using the Simoid function

Example; The activation (output) of layer two would be represented as follows.

$$ \overrightarrow a ^ {[2]} = [
    g(\overrightarrow W ^{[2]}_1 \cdot \overrightarrow a ^{[1]} \cdot b^{[2]}_1)
    \cdot \cdot \cdot
    g(\overrightarrow W ^{[2]}_{15} \cdot \overrightarrow a ^{[1]} \cdot b^{[2]}_{15})
    ] $$

The output of the third and last layer, tells us the probability.

$\overrightarrow a^{[3]}_1 \geq 0.5 $?

- Yes -> $\hat{y} = 1$ (The image is the digit "1")
- No -> $\hat{y} = 0$ (The image is not the digit "1")
