### Regression/Linear Model

The function implemented by a neuron with no activation is the same as in Course 1, linear regression:
$$ f_{\mathbf{w},b}(x^{(i)}) = \mathbf{w}\cdot x^{(i)} + b \tag{1}$$

We can define a layer with one neuron or unit and compare it to the familiar linear regression function.

```python
linear_layer = tf.keras.layers.Dense(units=1, activation = 'linear', )

# Examine the weights
linear_layer.get_weights()
```

```python
# There are no weights because they havent been instantiated.
# Run the model on one example to trigger the instantiation of the weights.

# Input needs to be 2D, so reshape
print(f"Reshaped {X_train[0].reshape(1,1)}")

a1 = linear_layer(X_train[0].reshape(1,1))
print("Output:")
print(a1)
```

The output is a "tensor" (aka an array) with a shape of (1,1).

```python
# The weights are randomply initialized to small numbers, and the bias defaults to 0 when initialized.
w, b= linear_layer.get_weights()
print(f"w = {w}, b={b}")
```

```python
# Replace the weights with know values
set_w = np.array([[200]])
set_b = np.array([100])

# set_weights takes a list of numpy arrays
linear_layer.set_weights([set_w, set_b])
print(linear_layer.get_weights())
```

```python
# Compare equation (1) to the layer output.

# Recalculate with new weights.
a1 = linear_layer(X_train[0].reshape(1,1))
print(a1)
alin = np.dot(set_w,X_train[0].reshape(1,1)) + set_b
print(f"Numpy prediction: {alin}")
```

They are the same value, now we can pass in the input vector to get all predictions.

```python
prediction_tf = linear_layer(X_train)
prediction_np = np.dot( X_train, set_w) + set_b

plt_linear(X_train, Y_train, prediction_tf, prediction_np)
```
