# Classification using linear regression

Answers are either true or false (0 or 1).

- Linear regression is not ideal for classification. An outlier datapoint can shift the decision boundary and sudenly flip several predictions.

```python
import numpy as np
import matplotlib.pyplot as plt
from lib.lab_utils_common import dlc, plot_data
from lib.plt_one_addpt_onclick import plt_one_addpt_onclick
plt.style.use('./lib/deeplearning.mplstyle')
```

```python
x_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])
X_train2 = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train2 = np.array([0, 0, 0, 1, 1, 1])
```

```python
# Define linear regression method

def perform_linear_regression(x_vect, weight, bias):

    result = []

    for i in range(len(x_vect)):

        input = x_vect[i]
        output = input * weight + bias
        result.append(output)
        
    return result
```

```python
# create predictions
predictions = perform_linear_regression(x_train, 0.20, 0)
print(predictions)
```

```python
pos = y_train == 1
neg = y_train == 0

fig,ax = plt.subplots(1,2,figsize=(8,3))
#plot 1, single variable
ax[0].scatter(x_train[pos], y_train[pos], marker='x', s=80, c = 'red', label="y=1")
ax[0].scatter(x_train[neg], y_train[neg], marker='o', s=100, label="y=0", facecolors='none', 
              edgecolors=dlc["dlblue"],lw=3)

ax[0].set_ylim(-0.08,1.1)
ax[0].set_ylabel('y', fontsize=12)
ax[0].set_xlabel('x', fontsize=12)
ax[0].set_title('one variable plot')
ax[0].legend()

ax[0].plot(x_train, predictions)

#plot 2, two variables
plot_data(X_train2, y_train2, ax[1])
ax[1].axis([0, 4, 0, 4])
ax[1].set_ylabel('$x_1$', fontsize=12)
ax[1].set_xlabel('$x_0$', fontsize=12)
ax[1].set_title('two variable plot')
ax[1].legend()
plt.tight_layout()
plt.show()
```

- On the one variable plot, 2.5 marks the `desicion boundary`, and it separates true and false values.
- If an outlier input were to be added, the predictions line would shift as well as the `descicion boundary`. This would have the unintended effect of flipping some values.
- On the `two variable plot`, there is no y axis because outputs can be represented with icons.

```python
x_train = np.array([0., 1, 2, 3, 4, 5, 10])
y_train = np.array([0,  0, 0, 1, 1, 1, 1])

pos = y_train == 1
neg = y_train == 0

fig,ax = plt.subplots(1,1,figsize=(8,3))

ax.scatter(x_train[pos], y_train[pos], marker='x', s=80, c = 'red', label="y=1")
ax.scatter(x_train[neg], y_train[neg], marker='o', s=100, label="y=0", facecolors='none', 
              edgecolors=dlc["dlblue"],lw=3)

ax.set_ylim(-0.08,1.1)
ax.set_ylabel('y', fontsize=12)
ax.set_xlabel('x', fontsize=12)
ax.set_title('one variable plot')
ax.legend()

predictions = perform_linear_regression(x_train, 0.10, 0)

ax.plot(x_train, predictions)
plt.show()
```

In this scenario, at least two negative values would now become positive.
