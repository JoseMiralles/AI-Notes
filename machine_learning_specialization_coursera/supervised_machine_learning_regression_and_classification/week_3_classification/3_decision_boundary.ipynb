# 1. Desicion Boundary Notes

## Formula

$$

f_{ \overrightarrow W \bullet \overrightarrow X }( \overrightarrow X ) =

g(z) = g(\overrightarrow W \bullet \overrightarrow X + b) =

{1 \over { 1 + e ^ {-z} }} =

{1 \over { 1 + e ^ {- (\overrightarrow W \bullet \overrightarrow X + b)} }}

$$

- **z** is the linear regression formula: $ z = \overrightarrow W \bullet \overrightarrow X + b $

All of this then equals to:

$$ g(z) = P(y=1|x;\overrightarrow W, b) $$

- $P(y=1|x;\overrightarrow W, b)$ is the probability that **y** is **1**.


## Threshold



If ${ \overrightarrow W \bullet \overrightarrow X }( \overrightarrow X ) \ge 0.5$ or $g(z) \ge 0.5$, then the prediction $\hat y = 1$. Otherwise, $\hat y = 0$.

- Notice in the chart above that if $z \ge 0$, then $g(z) \ge 0.5$
- And when $z \le 0$, then $g(z) \le 0.5 0$

## Linear descision boundary with two features

$$
g(z) = g(w_1 x_1 + w_2 x_2 + b)
$$

Since there are two features, these could be charted as follows:



The treshold in this case is when $x_1 + x_2 = 3$ (purple line).

$$
z = \overrightarrow W \bullet \overrightarrow X + b = 0
$$

$$
z = x_1 + x_2 - 3 = 0
$$


$$
x_1 + x_2 = 3
$$

```python
import numpy as np
%matplotlib widget
import matplotlib.pyplot as plt
from lib.lab_utils_common import plot_data, sigmoid, draw_vthresh
plt.style.use('./deeplearning.mplstyle')
```

### Dataset

- $\overrightarrow X$ is a numpy arraw with 6 training examples, each with 2 features.
- $\overrightarrow Y$ is also a numpy array with 6 examples. y is either 0 or 1.

```python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1]).reshape(-1,1) 
```

```python
fig,ax = plt.subplots(1,1,figsize=(4,4))
plot_data(X, y, ax)

ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.show()
```

```python
# Plot sigmoid(z) over a range of values from -10 to 10
z = np.arange(-10,11)

fig,ax = plt.subplots(1,1,figsize=(5,3))
# Plot z vs sigmoid(z)
ax.plot(z, sigmoid(z), c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
```

```python
## Plot desicion boundary
# Choose values between 0 and 6
x0 = np.arange(0,6)

x1 = 3 - x0
fig,ax = plt.subplots(1,1,figsize=(5,4))
# Plot the decision boundary
ax.plot(x0,x1, c="b")
ax.axis([0, 4, 0, 3.5])

# Fill the region below the line
ax.fill_between(x0,x1, alpha=0.2)

# Plot the original data
plot_data(X,y,ax)
ax.set_ylabel(r'$x_1$')
ax.set_xlabel(r'$x_0$')
plt.show()
```
