# Multiple Linear Regression

*Not to be confused with multivariate regression*

**Example:** Instead of only taking in *Size in feet*, the model can also take other variables.

| Size in feet$^2$ () | Bedrooms | Floors | Home age | **Price ($1,000s)** |
|--|--|--|--|--|
|$x_1$|$x_2$|$x_3$|$x_4$|**$y$**|
| 2104 | 5 | 1 | 45 | **460** |
| 1416 | 3 | 2 | 40 | **232** |
| ... |

- $n$ = the number of features or variables.
- $x_i$ = the $i^{th}$ feature.
- $\overrightarrow x^{(i)}$ = features of $i^{th}$ training example.
    - Ex: $\overrightarrow x^{(2)}$ = $[ 1416, 3, 2, 40 ]$ from the table above.
- $x_j^{(i)}$ = Is like accesing a specific value `x[j][i]` in a 2d array.

# Formula

Without vector notation:

$f_{w,b}(x) = w_1 x_1 + w_2 x_2 + w_3 x_3 + w_4 x_4 + b$

- There are multiple weights $\overrightarrow w^{(i)}$ just like there are multiple inputs $\overrightarrow x^{(i)}$.

With vector notation:

$f_{w,b}(x) = \overrightarrow w \bullet \overrightarrow x + b$

- "$\overrightarrow w \bullet \overrightarrow x$" is the **dot** product of $\overrightarrow w$ and $\overrightarrow x$
