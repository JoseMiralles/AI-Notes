# Feature engineering

The practice of coming up with features to add to a model.

Example: A model could include features for the frontage and the depth of real estate. However, an engineer may also want to add a new feature such as the area to see which ones of the three carries the most weight.

$$ f_{\overrightarrow w, b} (\overrightarrow x) = w_1 x_1 + w_2 x_2 + w_3 x_3 + b $$

- Where $w_1$ = frontage, $w_2$ = depth, and $w_3$ = area.
