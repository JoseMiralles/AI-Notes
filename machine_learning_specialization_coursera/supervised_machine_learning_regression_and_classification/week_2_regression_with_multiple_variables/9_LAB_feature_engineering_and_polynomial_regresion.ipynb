## Selecting features

Gradient descent helps in finding out if a feature is important compared to other features because it finds the weight of all features. A feature with a weight that is zero or close to zero could be seen as unimportant.

Example: $y=w_0x_0 + w_1x_1^2 + w_2x_2^3+b$

```python
# create target data
x = np.arange(0, 20, 1)
y = x**2

# engineer features .
X = np.c_[x, x**2, x**3]   #<-- added engineered feature
```

```python
model_w,model_b = run_gradient_descent_feng(X, y, iterations=10000, alpha=1e-7)

plt.scatter(x, y, marker='x', c='r', label="Actual Value"); plt.title("x, x**2, x**3 features")
plt.plot(x, X@model_w + model_b, label="Predicted Value"); plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.show()
```

In the output, ``w: [0.08 0.54 0.03]`` demonstrates that the second term holds a much higher weight than the other terms. Increasing the iterations for gradient descent would further increase the contrast.

After removing the "less important" terms, the formula goes back to: $$y = w_0 x_0^2 + b$$
