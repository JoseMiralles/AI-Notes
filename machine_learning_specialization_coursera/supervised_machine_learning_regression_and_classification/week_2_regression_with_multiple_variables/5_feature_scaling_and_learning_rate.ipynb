# 1.1 - Feature scaling

The goal is to widen or highten a contour plot to make it easier for gradient descent to complete. To acheive this, data ranges need to be scaled down to 0 - 1 by utilizing percentages.

Feature ranges may bee too large or too small in relation to the ranges of other features. This would require scaling.



### Examples:

To scale a feature, it can be divided by the max in range.

$x_1, scaled = {x_1 \over {2000}}$

$x_2, scaled = {x_2 \over 5}$

- Now $0.15 \le x_1, scaled \le 1$
- And $0 \le x_2, scaled \le 1$

## Mean Normalization

An approach in which ranges are reduced to be between -1 to 1.

$$ x_1 = {{ x_1 - μ_1 } \over { max - min }} $$

- $μ$ is the average.

## Z-score Normalization

This aproach utilizes the standard deviation and average to normalize values.

$$ x_1 = { {x_1 - μ_1} \over { \sigma _1 } } $$
