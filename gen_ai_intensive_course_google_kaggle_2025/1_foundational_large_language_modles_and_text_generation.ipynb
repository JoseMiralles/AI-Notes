# Foundational Large Language Models & Text Generation

## Prompt Engineering

- Well-structured prompts lead to better outputs, improving efficiency.

### Types of prompts

- **Zero-shot prompting**: Asking a question without examples.
- **One-shot prompting**: Providing one example to guide the model.
- **Few-shot prompting**: Giving multiple examples to set the response pattern.
- **Chain-of-thought prompting**: Encouraging step-by-step reasoning for complex problems.
