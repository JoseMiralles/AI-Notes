# Foundational Large Language Models & Text Generation

## Prompt Engineering

### Top-K Sampling

Top-K limits the model’s choices to the K most probable next tokens. Instead of considering all possible words, it selects from the top K highest-probability words.

Example:
A model is trying to complete:
"The cat sat on the..."

Without top-k, the model might consider all possible words based on their probabilities.
With top-k = 3, the model only considers the top 3 most likely words.

Probabilities (hypothetical):

- mat (0.4)
- floor (0.3)
- sofa (0.2)
- table (0.1)

If K=3, the model only picks from (mat, floor, sofa) and ignores table.


### Top-P (Nucleus) Sampling
Definition:
Top-P (a.k.a. nucleus sampling) picks from the smallest possible set of words whose cumulative probability exceeds P.

Example "The cat sat on the...":
If P=0.7, we sum probabilities until we exceed 0.7:

- mat (0.4) → included
- floor (0.3) → included (Total = 0.4 + 0.3 = 0.7 ✅)
- sofa (0.2) → ❌ ignored
- table (0.1) → ❌ ignored

Now, the model randomly samples from (mat, floor) but not sofa or table.
