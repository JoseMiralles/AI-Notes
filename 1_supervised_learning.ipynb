# Supervised Learning

- Learns from data with provided "right" answers. 
- Rapid advancement 
- Most used in real world applications 
- X to Y, or input to output.

Examples: Spam filter, audio to text, English to Spanish.


## Two types of supervised learning algorithms 


1. Regression 

- like predicting a house price given its square footage. 

- It has an infinite possible number of outputs. 

 

2. Classification 

- Predicts categories, and can output a small number of possible outputs.
